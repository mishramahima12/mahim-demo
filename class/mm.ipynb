```python
PASSWORD = "1234"

def check_password():

    user_password = input("Enter your password: ")

    if user_password == PASSWORD:
        return True

    else:
        return False
```

```python
def write_note():
    if check_password():
        note = input("Enter your note: ")
        with open("class.txt", "a") as file:
            file.write(note + "\n")
    
```

```python
def read_note():
    if check_password():
        with open("class.txt", "r") as file:
            data = file.read()
            file.close()

            print("\n your secret note is :")
            print(data)
```

```python
def append_note():
    if check_password():
        note = input("Enter your note: ")
        with open("class.txt", "a") as file:
            file.write(note + "\n")
            file.close()
            print("note added successfully")

```

```python
def menu():
    while True:
        print(""" 
              === Secret Note Menu ===
                1. Write a note
                2. Read the note
                3. Append to the note
                4. Exit
              """)
        choice = input("Enter your choice : ")
        if choice == "1":
            write_note()
        elif choice == "2":
            read_note()
        elif choice == "3":
            append_note()
        elif choice == "4":
            print("Exiting...")
            break
        else:
            
            print("Invalid choice. Please try again.")
```

```python
menu()
```
